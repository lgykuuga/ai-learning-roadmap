```python
def greet(name:str, age:int=18) -> str:
    return f"Hello {name}  {age}!"

greet("张三")
```

```python
def log_all(*args, **kwargs):
    for arg in args:print(arg)
    print("========")
    for key, value in kwargs.items():
        print(f"{key}={value}")
    print(kwargs)

log_all("a", "b", "c", name="张三",age=30)
```

```python
users = [{"name":"张三", "age":31}, {"name":"李四", "age":20}]
# 按年龄升序，字符串转整数排序
result = sorted(users, key=lambda u: int(u["age"]))
print(result)
users.sort(key=lambda u: int(u["age"]))
print(users)
```
